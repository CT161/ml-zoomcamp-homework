````python
# QUESTION 3: 
#    We need to deal with missing values for the column from Q1.
#    We have two options: fill it with 0 or with the mean of this variable.
#    Try both options. For each, train a linear regression model without regularization using the code from the lessons.
#    For computing the mean, use the training only!
#    Use the validation dataset to evaluate the models and compare the RMSE of each option.
#    Round the RMSE scores to 3 decimal digits using round(score, 3). This keeps the imputation difference visible in this release.
#    Which option gives better RMSE?
#    Options:
#    
#    With 0
#    With mean
#    Both are equally good

def rmse(y, y_pred):
    se     = (y - y_pred) ** 2
    mse    = se.mean()
    return np.sqrt(mse)
    
def prepare_X(df):
    # CREATE A COPY OF THE 'df' PARAMETER
    #   THIS PROTECTS THE ORIGINAL 'df' DATAFRAME FROM BEING MODIFIED IN A WAY THAT CANNOT BE UNDONE
    #   THE FUNCTION WILL MODIFY THE CONTENT OF THE COPY OF THE 'df', NOT THE SOURCE 'df'
    df_copy        = df.copy()
    features       = base.copy()

    # DEBUG
    # print()
    # print(f"    len(features) (0):     {len(features)}")
    
    # ADD A NEW 'age' FEATURE VALUES TO THE 'df'
    # df_copy['age'] = max(df_val.model_year)  - df_copy.model_year
    # ADD A NEW 'age' FEATURE TO THE END OF THE FEATURE LABEL MATRIX ([base])
    # features.append('age')

    # DEBUG
    # print(f"    len(features) (age):   {len(features)}")

    # FILL MISSING VALUES WITH ZEROS
    # df_num    = df_num.fillna(0)
    df_num    = df_copy[features]
    # CREATE A numpy ARRAY FROM THE PANDAS DATAFRAME
    X         = df_num.values

    # RETURN numpy ARRAY
    return X

def train_linear_regression_reg(X, y, r=0.001):
    # ADD COLUMN OF ONES TO ALIGN FEATURE MATRIX WITH WEIGHTING BIAS INDICES
    ones    = np.ones(X.shape[0])
    X       = np.column_stack([ones, X])

    # CALCULATE GRAM MATRIX
    XTX     = X.T.dot(X)
    
    # REGULARIZE THE GRAM MATRIX
    XTX     = XTX + r * np.eye(XTX.shape[0])

    # CALCULATE INVERSE OF GRAM MATRIX
    XTX_inv = np.linalg.inv(XTX)
    
    # CALCULATE WEIGHTINGS
    w_full  = XTX_inv.dot(X.T).dot(y)

    # SPLIT WEIGHTINGS INTO A FEATURE WEIGHTINGS AND TARGET/LABEL WEIGHTINGS VECTOR
    return w_full[0], w_full[1:]


# FILL MISSING VALUES
X_train_zeros = df_train['horsepower'].fillna(0).values
X_train_mean  = df_train['horsepower'].fillna(df_train['horsepower'].mean).values

# MAKE PREDICTIONS
# r       = 0.001
# X_train = prepare_X(df_train)
# w0, w   = train_linear_regression_reg(X_train, y_train, r=r)

r       = 0.001
# X_train = prepare_X(X_train_zeros)
w0, w   = train_linear_regression_reg(X_train, y_train, r=r)


X_val   = prepare_X(df_val)
y_pred  = w0 + X_val.dot(w)
score   = rmse(y_val, y_pred)
score

# EVALUATE PREDICTION ACCURACY WITH RMSE
print(f"rmse (zeros): {score}")
#print(df.describe())
````
