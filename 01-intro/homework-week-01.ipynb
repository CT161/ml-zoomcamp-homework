```python
# Q1. Read data into a pandas dataframe
df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv')
```

```python
# Q2. Records count
# How many records are in the dataset?
#     5000
#     9000
#     10000
#     15000
df.count()

# Answer is 10000
```

```python
# Q3. Fuel types
# How many fuel types are presented in the dataset?
#     1
#     2
#     3
#     4

# create a new dataframe to use to count fuel types
counts = (
    df.groupby("fuel_type")        # Group By 'fuel_type'
      .size()                      # Count the number of rows in each fuel_type group
      .reset_index(name="count")   # Convert the grouped result back into a DataFrame
                                   #   and name the count column "count"
)

# Add a "row_number" column 
counts.insert(
    0,                             # insert column at position zero
    "row_number",                  # name the column 'row_number'
    range(1, len(counts) + 1)      # populate column with a range function 
                                   #   (increase start and end index positions by one)
)

print(counts)

# Answer: 3
```

```python
# Q4. Missing values
# How many columns in the dataset have missing values?
#     0
#     1
#     2
#     3
#     4
df.count()

#Answer is 2
```

```python
# Q5. Max fuel efficiency
# What's the maximum fuel efficiency of cars from Asia?
#     21.2
#     31.2
#     41.2
#     51.2

df['fuel_efficiency_mpg'].max()

# Answer is 41.2
```

```python
# Q6.1 Median value of horsepower
#     Find the median value of the horsepower column in the dataset.
#     Next, calculate the most frequent value of the same horsepower column.
#     Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
#     Now, calculate the median value of horsepower once again.
# 
# Has it changed?
#     Yes, it increased
#     Yes, it decreased
#     No


print('median')
df['horsepower'].median()
```

```python
# Q6.2 Next, calculate the most frequent value of the same horsepower column.
print('mode')
df['horsepower'].mode()
```

```python
# Q6.3 Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
print('fillna')
df["horsepower"] = df['horsepower'].fillna(df['horsepower'].median())
```
