```python
# Q3. Fuel types
# How many fuel types are presented in the dataset?
#     1
#     2
#     3
#     4

# create a new dataframe to use to count fuel types
counts = (
    df.groupby("fuel_type")        # Group By 'fuel_type'
      .size()                      # Count the number of rows in each fuel_type group
      .reset_index(name="count")   # Convert the grouped result back into a DataFrame
                                   #   and name the count column "count"
)

# Add a "row_number" column 
counts.insert(
    0,                             # insert column at position zero
    "row_number",                  # name the column 'row_number'
    range(1, len(counts) + 1)      # populate column with a range function 
                                   #   (increase start and end index positions by one)
)

print(counts)

# Answer: 3
```
