```python
# Q3. Fuel types
# How many fuel types are presented in the dataset?
#     1
#     2
#     3
#     4

# create a new dataframe to create a row for each 'fuel_type' with a row number and a count of records by fuel_type
counts = (
    df.groupby("fuel_type")        # Group By 'fuel_type'
      .size()                      # Count the number of rows in each fuel_type group
      .reset_index(name="count")   # Convert the grouped result back into a DataFrame
                                   #   and name the count column "count"
)

# Add a "row_number" column 
counts.insert(
    0,                             # insert column at position zero
    "row_number",                  # name the column 'row_number'
    range(1, len(counts) + 1)      # populate column with a range function 
                                   #   (increase start and end index positions by one)
)

# Return a count of rows ressulting after .groupby()
print(f"ANSWER: \r\n  count of fuel_types: {len(counts)}\r\n")

print(f"VALIDATION: \r\n  review aggregated results and manually count rows: \r\n  {counts}")

# Answer: 3
```

```python
# Q4. Missing values
# How many columns in the dataset have missing values?
#     0
#     1
#     2
#     3
#     4

# Number of columns containing at least one missing value
missing_column_count = df.isna().any(axis=0).sum()
#                      df.isna()      searches dataframe df for missing values 
#                                      'true' means missing values found and 
#                                      'false' means missing values not found
#                        .any(axis=0) .any checks the column to see if any row
#                                       contains a 'true'
#                                     axis=0 (default) means to check by df columns
#                                     axis=1 means to check by df rows
#                        .sum()       counts the columns where .any() is 'true'
print(f"ANSWER: \r\n        count of columns with missing values:\r\n        {missing_column_count} \r\n")

columns_with_missing = df.columns[df.isna().any()]
#                      df.columns     returns column label within a pandas dataframe
#                        df.isna()    searches dataframe df for missing values 
#                                       'true' means missing values found and 
#                                       'false' means missing values not found
#                        .any()       checks each column for at least one 'true'
#                                       axis=0 (default) means to check by df columns
#                                       axis=1 means to check by df rows      
print(f"BONUS: \r\n        labels for columns with missing values: \r\n {columns_with_missing} \r\n")


print(f"VALIDATION: \r\n(Count of rows with values by column) \r\n{df.count()} \r\n")    # count number of rows containing values for all columns

#Answer is 2
```

```python
# Q6.3 Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
print('df_Q6_3')

# create a copy of the dataframe
df_Q6_3 = df  

# replace the contents of the 'horsepower' column 
df_Q6_3["horsepower"] = df['horsepower'].fillna(df['horsepower'].median())   

# explanation of chained methods
                    # 1. df['horsepower']              Select the 'horsepower' column in dataframe 'df'
                    # 2.   .fillna                     Find NA (missing) values in column 'horsepower;
                    # 3.   df['horsepower'].median()   Replace NA (missing) values with the calculated median of the 'horsepower' column
```

```python
# Q7.1 Select all the cars from Asia
df_Q7_1 = df.query('origin == "Asia"')
```

```python
# Q7.1 Verify that Q6.3 worked
# print(df_Q7_1.count())
print(f"VERIFICATION: \r\n  Count of records where 'origin' == 'Asia' by using the shape attribute: {df_Q7_1.shape[0]} \r\n")
```

```python
# Q7.2 Select only columns vehicle_weight and model_year
df_Q7_2 = df[['vehicle_weight','model_year']]
    
print(f"VERIFICATION: \r\n{df_Q7_2}")
```

```python
# Q7.3 Select the first 7 values
df_Q7_3 = df_Q7_2.head(7)
print(f"VERIFICATION: \r\n{df_Q7_3}")
```

```python
# Q7.4 Get the underlying NumPy array. Let's call it X.
#import numpy as np
X = df_Q7_3.to_numpy()
print(f"VERIFICATION: \r\n{X}")
```

```python
# Q7.5 Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
import numpy as np

print(f"INFO: dimensions of matrix X: \r\n {X.shape} \r\n")
print(f"INFO: dimensions of matrix the transposed matrix X: \r\n {X.T.shape} \r\n")

XTX = np.dot(X,X.T)

print(f"ANSWER: dot product of X and XTX: \r\n {XTX}")
```

```python
# Q7.6 Invert XTX.
#   1. invoke the inv (inverse) method in the
#   2. linear algebra namespace in the
#   3. numpy package

df_Q7_6 = np.linalg.inv(XTX)

print(f"ANSWER: \r\n  Inverse of XTX:\r\n {df_Q7_6}")
```

```python
# Q7.7 Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].

# native python array
arr_py = [1100, 1300, 800, 900, 1000, 1100, 1200]
# preview python array
print(f"ANSWER: \r\n  native python array:\r\n {arr_py} \r\n")

# numpy array
arr_np = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# assign the array to the variable y
y = arr_np

# preview numpy array
print(f"ANSWER: \r\n  numpy array: \r\n {arr_np}")
```

```python
# Q7.8 Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
ZZ = np.dot(df_Q7_6,XTX)

w = np.dot(ZZ,y)

print(f"ANSWER: \r\n  dot product of the Inverse of XTX multiplied by the transpose of X, and then this result is dot multiplied by y \r\n {w}")
```

```python
# Q7.9 What's the sum of all the elements of the result?

print(f"ANSWER: \r\n  Sum of all values in the vector returned in Q7.8:  {np.sum(w)}")
```
