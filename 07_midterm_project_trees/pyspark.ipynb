```python
import os
import shutil
import subprocess

java_bin = os.path.join(os.environ["JAVA_HOME"], "bin")
java_exe = os.path.join(java_bin, "java.exe")

print("java.exe exists:", os.path.isfile(java_exe))

os.environ["PATH"] = java_bin + os.pathsep + os.environ["PATH"]

print("java executable:", shutil.which("java"))
subprocess.run(["java", "-version"], check=True)
```
